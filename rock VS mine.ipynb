```python
x = df.drop(columns = [60])
y = df[60]
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,stratify=y,random_state=RANDOM_STATE)
```

```python
print("dataset shape",x.shape)
print("dataset shape",x_train.shape)
print("dataset shape",x_test.shape)
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
x_test_scaled
```

```python
from sklearn.svm import SVC 
```

```python
model = SVC()
```

```python
model.fit(x_train_scaled,y_train)
```

```python
y_train_pre = model.predict(x_train_scaled)
train_acc = accuracy_score(y_train,y_train_pre)
print(f"Training data accuracy:{round(train_acc *100,2)}%")
```

```python
y_test_pre = model.predict(x_test_scaled)
test_acc = accuracy_score(y_test,y_test_pre)
print(f"Testing data accuracy:{round(test_acc *100,2)}%")
```

```python
print(" traing data classification report")
print(classification_report(y_train,y_train_pre))
```

```python
print(" testing data classification report")
print(classification_report(y_test,y_test_pre))
```

```python
def predict_object(input_features):
    # scale features
    scaled_features = scaler.transform([input_features])
    # get prediction from teh model
    prediction = model.predict(scaled_features)
    print("Model prediction", prediction)
    if prediction[0] == 1:
        print("The object is identified as Mine 💣")
    else:
        print("The object is identified as Rock 🪨")
```

```python
test_1 = x_test.loc[103].tolist()
print(test_1)
```
